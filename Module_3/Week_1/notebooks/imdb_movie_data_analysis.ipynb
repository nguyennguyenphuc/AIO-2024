```python
import sys
import os

# Thêm đường dẫn đến thư mục gốc của dự án
sys.path.append(os.path.abspath('..'))
```

```python

# Import các hàm từ các file
from src.imdb_movie_data.data_loading import load_data, view_data
from src.imdb_movie_data.data_cleaning import check_missing_values, drop_missing_values, fill_missing_values
from src.imdb_movie_data.data_analysis import select_columns, filter_data, group_by_column, sort_data, apply_custom_function
from src.imdb_movie_data.utils import display_missing_values, rating_group
```

```python
# 1. Đọc dữ liệu từ file CSV
file_path = '../data/imdb_movie_data/IMDB-Movie-Data.csv'
df = load_data(file_path)
# 2. Xem qua dữ liệu để có cái nhìn tổng quan
view_data(df)
```

```python
# 3. Hiểu các thông tin cơ bản về dữ liệu
print("Shape of the dataset:", df.shape)
print(df.info())
print(df.describe())
```

```python
print(df.head())
```

```python
# 4. Chọn và cắt dữ liệu (Indexing and Slicing)
# Chọn cột 'title' và 'Director'
title_director = select_columns(df, ['Title', 'Director'])
view_data(title_director)
# Chọn dòng từ 10 đến 20 của dữ liệu
subset_df = df[10:21]
view_data(subset_df)

```

```python
# 5. Lọc dữ liệu dựa trên điều kiện (Conditional filtering)
# Lọc các phim có điểm IMDB trên 8.0
high_rated_movies = filter_data(df, df['Rating'] > 8.0)
view_data(high_rated_movies[['Title', 'Rating']])
# Lọc các phim của đạo diễn Christopher Nolan
nolan_movies = filter_data(df, df['Director'] == 'Christopher Nolan')
view_data(nolan_movies[['Title', 'Director']])
```

```python
# 6. Thực hiện các phép toán nhóm (Groupby operations)
# Nhóm dữ liệu theo đạo diễn và tính trung bình điểm IMDB cho mỗi đạo diễn
director_group = group_by_column(df, 'Director', 'Rating')
view_data(director_group)
# Nhóm dữ liệu theo năm sản xuất và tính tổng doanh thu
year_revenue_group = group_by_column(df, 'Year', 'Revenue (Millions)', 'sum')
view_data(year_revenue_group)
```

```python
# 7. Sắp xếp dữ liệu (Sorting operation)
# Sắp xếp các phim theo doanh thu từ cao đến thấp
sorted_by_gross = sort_data(df, 'Revenue (Millions)', ascending=False)
view_data(sorted_by_gross[['Title', 'Revenue (Millions)']])
# Sắp xếp các đạo diễn theo điểm trung bình IMDB từ cao đến thấp
sorted_directors = sort_data(director_group, 'Rating', ascending=False)
view_data(sorted_directors)
```

```python
# 8. Xem các giá trị bị thiếu (View missing values)
display_missing_values(df)
```

```python
# 9. Xử lý giá trị thiếu bằng cách xóa bỏ (Deal with missing values - Deleting)
df_dropped_na = drop_missing_values(df)
print("Shape after dropping rows with missing values:", df_dropped_na.shape)
```

```python
# 10. Xử lý giá trị thiếu bằng cách điền vào (Deal with missing values - Filling)
df_filled = fill_missing_values(df)
display_missing_values(df_filled)
```

```python
# 11. Sử dụng hàm apply để thực hiện các phép tính cụ thể (Apply() functions)
# Tạo một cột mới 'profit' dựa trên gross và budget
df_filled = apply_custom_function(df_filled, lambda row: rating_group(row['Rating']), 'Rating Category')
view_data(df_filled[['Title', 'Revenue (Millions)', 'Rating Category']])
```
